# `misc_widgets` — reusable Jupyter widgets

This notebook demonstrates the standalone widgets provided by
`evbeantools.juptools.misc_widgets`.

These widgets are designed to be reusable building blocks that can be
embedded in any Jupyter-based UI, independently of the Sankey / financial-flow
diagrams.

## `DateRangeWidget`

A composite widget (subclass of `ipywidgets.HBox`) that bundles two DatePickers and a SelectionRangeSlider into a single, self-contained object.

| Feature | Details |
|---|---|
| **Displayable** | It's a widget — just call `display(drw)` |
| **Observable** | Has a `value` trait (tuple of start/end dates) — use `drw.observe(cb, names="value")` |
| **Properties** | Read `drw.start_date` / `drw.end_date` as shortcuts for `drw.value[0]` / `drw.value[1]` |

All three internal sub-widgets are kept in sync automatically: dragging the slider updates the pickers and vice-versa.

```python
import datetime
from IPython.display import display

from evbeantools.juptools import DateRangeWidget

import plotly.io as pio
```

### Basic usage

Supply a `min_date` and `max_date` to define the selectable range, then
call `display()` on the returned widget.

```python
min_date = datetime.date(2023, 1, 1)
max_date = datetime.date(2025, 12, 31)

drw = DateRangeWidget(min_date, max_date)

display(drw)
```

### Reading the selected range

After interacting with the widget above, run the cell below to inspect the
currently selected dates. `None` means the user has not yet picked a value
(i.e. "no constraint").

```python
print(f"value:      {drw.value}")
print(f"Start date: {drw.start_date}")
print(f"End date:   {drw.end_date}")
```

### Reacting to changes with an observer

You can observe the `value` trait — exactly the same pattern as any other
ipywidgets widget: `drw.observe(callback, names="value")`.

```python
import ipywidgets as ipw

min_date = datetime.date(2024, 1, 1)
max_date = datetime.date(2024, 12, 31)

drw2 = DateRangeWidget(min_date, max_date)

log = ipw.Output()

def on_date_change(change):
    with log:
        log.clear_output()
        s, e = change["new"]
        if s and e:
            delta = (e - s).days
            print(f"Selected range: {s.isoformat()} to {e.isoformat()}  ({delta} days)")
        else:
            print(f"Start: {s}, End: {e}")

drw2.observe(on_date_change, names="value")

display(drw2, log)
```
